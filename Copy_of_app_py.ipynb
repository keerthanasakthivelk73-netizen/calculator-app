```python
 %%writefile app.py
import streamlit as st

st.title("Calculator")

num1 = st.number_input("First Number")
num2 = st.number_input("Second Number")

operation = st.selectbox("Operation", ["+", "-", "*", "/"])

if st.button("Calculate"):
    if operation == "+":
        st.success(num1 + num2)
    elif operation == "-":
        st.success(num1 - num2)
    elif operation == "*":
        st.success(num1 * num2)
    elif operation == "/":
        if num2 != 0:
            st.success(num1 / num2)
        else:
            st.error("Cannot divide by zero")
```
